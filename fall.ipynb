```python
from ultralytics import YOLO

# Check version
import ultralytics
ultralytics.checks()
```

```python
%%writefile fall.yaml
train: fall_dataset/train/images
val: fall_dataset/valid/images
test: fall_dataset/test/images

nc: 1
names: ['Fall-Detected']

```

```python
model = YOLO('yolov8n.pt')
```

```python
results = model.train(
    data='fall.yaml',
    epochs=50,           # increase if needed
    imgsz=640,           # image size
    batch=16,            # depends on your GPU
    name='fall_detection_yolov8',
    project='runs/train',
)
```

```python
model.val(data='fall.yaml', split='test')
```

```python
model.predict(source='data/test/non-fall(28).avi', show=True, save=True, project='outputs', name='my_video')
```
